```python
import numpy as np
from sklearn.impute import SimpleImputer

# 1. Define the updated 3x3 matrix
matrix = [
    [1, 2, np.nan],
    [4, np.nan, 6],
    [np.nan, 8, 9]
]

# 2. Initialize and run the imputer
imputer = SimpleImputer(strategy='mean')
imputed_matrix = imputer.fit_transform(matrix)

# 3. Print the result
print(imputed_matrix)

```
